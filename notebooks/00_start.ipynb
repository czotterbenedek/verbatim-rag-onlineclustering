```python
from dotenv import find_dotenv, load_dotenv
import os

load_dotenv(find_dotenv(), override=True)
os.environ.pop("OPENAI_API_KEY", None)

from verbatim_rag import VerbatimIndex, VerbatimRAG
from verbatim_core import VerbatimTransform
from verbatim_core.llm_client import LLMClient
```

```python
import os

model = os.getenv("LLM_MODEL", "llama3.2:3b")
api_base = os.getenv("LLM_API_BASE", "http://localhost:11434/v1")
api_key = os.getenv("LLM_API_KEY", "ollama")
llm_client = LLMClient(model=model, api_base=api_base, api_key=api_key)
vt = VerbatimTransform(llm_client=llm_client)

response = vt.transform(
    question="What is Verbatim RAG, how does its evidence extraction approach work, and what problem is it designed to solve?",
    context=[
        {
            "content": """
            Verbatim RAG is a retrieval-augmented generation approach that
            focuses on extracting precise evidence from retrieved documents
            before generating an answer. The system aims to improve the
            faithfulness and traceability of generated answers by grounding
            them in specific textual evidence.
            """,
            "title": "Verbatim RAG documentation"
        },
        {
            "content": """
            The Verbatim RAG approach separates retrieval from evidence
            extraction. After retrieving relevant passages, an evidence
            extraction step identifies the portions of those passages that
            directly support the answer. The extracted evidence is then used
            to generate a grounded response.
            """,
            "title": "Verbatim RAG paper"
        }
    ]
)

print(response.answer)
```
